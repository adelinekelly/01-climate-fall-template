*  Warming temperatures caused by climate change are causing increases in bedrock weathering in high-elevation areas such as Green Lakes Valley (part of the Niwot Ridge LTER, close to Boulder, CO). This is in turn leading to higher concentrations of sulfate and base cations such as calcium in alpine streams (Crawford et al. 2019). 
* Climate change is leading to notable increases in wildfire activity across the Western U.S. Paleoecological evidence indicates that within the 21st century, subalpine forests of the Southern Rocky Mountains (including Colorado) have burned at a rate that is nearly double that of the past 2,000 years (Higuera et al. 2021). 
* Colorado's April 1 SWE (snow water equivalent) value for 2026 was the lowest recorded since SNOTEL monitoring began in the 1980s. This was due largely to far warmer than average temperatures and has lead to prolonged drought conditions (Drought.gov 2026)
